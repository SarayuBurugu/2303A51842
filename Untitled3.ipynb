1. Arithmetic Mean Method


```python
n=int(input("Enter number of machines: "))
machines={}
for i in range(n):
    name=input("Enter machine name: ")
    rewards=list(map(int,input("Enter rewards: ").split()))
    machines[name]=rewards

best=None
high=float('-inf')

for name,rewards in machines.items():
    q=sum(rewards)/len(rewards)
    print(f"Machine {name}")
    print(f"Rewards: {rewards}")
    print(f"Q({name}) = {q:.2f}")
    if q>high:
        high=q
        best=name

print("Greedy Selected Machine:",best)
```

2. Incremental Update Method

```python
n=int(input("Enter number of machines: "))
q=[0]*n
count=[0]*n
names=[]

for i in range(n):
    name=input("Enter machine name: ")
    names.append(name)
    rewards=list(map(int,input("Enter rewards: ").split()))
    print(f"\nMachine {name}")
    for r in rewards:
        count[i]+=1
        q[i]=q[i]+(r-q[i])/count[i]
        print(f"Reward: {r}, N={count[i]}, Q={q[i]:.2f}")
    print(f"Final Q({name}) = {q[i]:.2f}")

best=q.index(max(q))
print("\nGreedy Selected Machine:",names[best])
```
