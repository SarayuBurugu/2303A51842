```python
import random

# ==========================================
# EPSILON-GREEDY MULTI-ARMED BANDIT
# ==========================================

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2

# Part B - Initialize N, R and Q tables
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': 0, 'B': 0, 'C': 0}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

# Given reward patterns
reward_pattern = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Make random results reproducible
random.seed(15)

print("=== Epsilon-Greedy Multi-Armed Bandit ===")
print("Machines:", machines)
print("Number of plays:", N_play)
print("Epsilon:", epsilon)

# ==========================================
# 10 PLAYS
# ==========================================

for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Part C - Epsilon-Greedy Decision
    if r < epsilon:
        # Exploration
        machine = random.choice(machines)
        decision = "Exploration"
    else:
        # Exploitation
        max_q = max(Q.values())

        # If more than one machine has same Q-value,
        # randomly select one of them
        best_machines = [
            m for m in machines
            if Q[m] == max_q
        ]

        machine = random.choice(best_machines)
        decision = "Exploitation"

    print("\nPlay", play)
    print("Random number:", round(r, 4))
    print("Decision:", decision)
    print("Selected machine:", machine)

    # ======================================
    # Part C - Reward Collection
    # ======================================

    # Get the next reward from the given pattern
    reward = reward_pattern[machine][N[machine]]

    print("Enter reward:", reward)

    # ======================================
    # Part D - Q-Value Update
    # ======================================

    # Update number of times machine is selected
    N[machine] += 1

    # Update total reward
    R[machine] += reward

    # Incremental sample-average formula
    Q[machine] = Q[machine] + (
        reward - Q[machine]
    ) / N[machine]

    print("Updated N:", N)
    print("Updated R:", R)
    print("Updated Q:", Q)


# ==========================================
# Part E - FINAL RESULTS
# ==========================================

print("\n" + "=" * 50)
print("FINAL RESULTS")
print("=" * 50)

print("\nMachine\tN (Count)\tR (Total Reward)\tQ (Estimated Value)")

for machine in machines:
    print(
        machine,
        "\t",
        N[machine],
        "\t\t",
        f"{R[machine]:.2f}",
        "\t\t\t",
        f"{Q[machine]:.2f}"
    )

# ==========================================
# ANALYSIS
# ==========================================

print("\nAnalysis:")

for machine in machines:
    print(
        f"Machine {machine}: "
        f"Played {N[machine]} times, "
        f"total reward = {R[machine]:.2f}, "
        f"estimated Q-value = {Q[machine]:.2f}"
    )

best_machine = max(Q, key=Q.get)

print("\nBest machine based on the final Q-value:",
      best_machine)
```
